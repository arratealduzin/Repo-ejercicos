# **CONTRASTES DE HIPÓTESIS**

- Técnica de estadística inferencial -> relación de variables de forma matemática. (Módulo de ML)
- Estadística descriptiva -> relación de variables de forma visual.

## **Coeficiente de correlación ≠ p-valor**

1. ***Coeficiente de correlación***

Mide la **fuerza** y la **dirección** de la relación entre variables. (imagen)

Rango: -1 a 1

--> Cuanto más cerca de -1 o 1, más fuerte es la relación  
--> **NO** nos dice si la relación es significativa, solo cuán fuerte es.

2. ***P-Valor***

Mide si la relación observada podría deberse al azar.

Rango: 0 a 1

--> No mide fuerza, mide evidencia estadística  
--> Cuanto más cerca de 0, más evidencia contra la hipótesis nula


*Resumen*:
- El **coeficiente** dice “cuánta relación hay”.  
- El **p-valor** dice “si esa relación es estadísticamente fiable”.

## **Tipos de Hipótesis**

**Hipótesis nula** -> SIEMPRE va a ser que las variables que estamos investigando van a ser INDEPENDIENTES, es decir, NO TIENEN RELACIÓN.

**Hipótesis alternativa** -> Las variables SÍ tienen relación, es decir, son DEPENDIENTES
