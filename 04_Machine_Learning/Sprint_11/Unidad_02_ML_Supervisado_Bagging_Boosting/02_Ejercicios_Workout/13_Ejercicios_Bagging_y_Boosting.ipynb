````python
import matplotlib.pyplot as plt
import numpy as np

def plot_mnist_images(images, bck_white = False):
    """
    Función para mostrar imágenes del dataset MNIST.

    :param images: Array de numpy con múltiples imágenes, cada una de 784 elementos.
    """
    # Número de imágenes
    num_images = len(images)

    # Crear una cuadrícula de subplots
    cols = int(np.sqrt(num_images))
    rows = (num_images // cols) + 1

    fig, axes = plt.subplots(rows, cols, figsize=(cols*2, rows*2))
    axes = axes.flatten()

    for i, ax in enumerate(axes):
        if i < num_images:
            # Convertir cada imagen de 784 elementos a un array 2D de 28x28
            img_matrix = images[i].reshape(28, 28)

            # Mostrar la imagen
            ax.imshow(img_matrix, cmap='gray' if not bck_white else "binary" )

            # Quitar los ejes
            ax.axis('off')
        else:
            ax.axis('off')

    plt.tight_layout()
    plt.show()

````

````python
plot_mnist_images(images)
````

¿Cuántos pixeles tiene de alto y ancho cada imagen?

Cada imagen de MNIST es una imagen cuadrada pequeña de 28×28 píxeles, que es por eso que los datos originales tienen 784 columnas (28 × 28 = 784) para representar todos los píxeles de la imagen.  
Cada número del array = 1 píxel  
784 números = 784 píxeles (una imagen completa de 28×28)  
El valor del número = qué tan claro/oscuro es ese píxel.

¿Importa qué número elijas? No, no hay una ventaja matemática en usar 42 sobre 0 o 100. Lo importante es que el valor sea fijo para que tus experimentos sean consistentes.  
La recomendación es utilizar el random_state con números fijos solo para temas de pruebas, pero en producción debemos omitir su uso ya que así obtenemos un resultado diferente en cada ejecución y es lo que deseamos.

### Ejercicio 5

Obten la feature importance del modelo pero muéstrala como una imagen para que veamos que píxeles son más importantes (hay tantos coeficientes como píxeles, recuerda). PISTA: Igual puedes usar parte del código de la función anterior o todo.

````python
fea_imp = arbol_clas.feature_importances_
fea_imp.shape
````

Estás intentando usar la función plot_mnist_images() para visualizar las importancias de las características (feature_importances_), pero esa función está diseñada para mostrar imágenes completas de dígitos MNIST (vectores de 784 elementos que representan píxeles).
Las feature_importances_ son 784 valores individuales (uno por cada píxel), que son un vector, donde cada valor indica qué tan importante es ese píxel para la clasificación. No son imágenes de dígitos, sino un "mapa de calor" de importancia.

````python
# Convertir el vector de 784 elementos a una matriz 28x28
importance_matrix = fea_imp.reshape(28, 28)

# Crear el mapa de calor
plt.figure(figsize=(8, 6))
plt.imshow(importance_matrix, cmap='coolwarm', interpolation='nearest')
plt.colorbar(label='Importancia')
plt.title('Importancia de cada píxel en la clasificación')
plt.xlabel('Píxel X')
plt.ylabel('Píxel Y')
plt.show()
````

**imshow()** es una función de Matplotlib, que significa "image show" (mostrar imagen).  
¿Qué hace?  
Muestra datos numéricos en forma de imagen o mapa de calor, donde cada valor numérico se representa con un color.  
Sintaxis básica:  
```python
 plt.imshow(datos, cmap='nombre_del_mapa_de_colores') 
```  
datos: Una matriz 2D (como 28x28) con valores numéricos,  
interpolation: Cómo suavizar la imagen.  

X e Y son simplemente las coordenadas de posición de cada píxel en la imagen de 28x28.
